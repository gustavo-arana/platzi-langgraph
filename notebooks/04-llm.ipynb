```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=1.0,  # Gemini 3.0+ defaults to 1.0
    max_tokens=None,
    timeout=None,
    max_retries=2,
)

response = model.invoke("Hola, como estas?")
response.pretty_print()
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.7-flash",
    temperature=1.0,  # Gemini 3.0+ defaults to 1.0
    max_tokens=None,
    timeout=None,
    max_retries=2,
)

response = model.invoke("Hola, como estas?")
response.pretty_print()
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

msg_1 = SystemMessage(content="You are a helpful assistant that talks in Spanish.")
msg_2 = HumanMessage(content="Hola, me llamo Juan")
msg_3 = AIMessage(content="Hola Juan, como estas?")
msg_4 = HumanMessage(content="Como me llamo?")

history = [msg_1, msg_2, msg_3, msg_4]
response = model.invoke(history)
response.pretty_print()
```

```python
from langchain_anthropic import ChatAnthropic

llm_ant = ChatAnthropic(
    model="claude-sonnet-4-5-20250929"
)

llm_ant_response = llm_ant.invoke(history)
llm_ant_response.pretty_print()
```

```python
from langchain.chat_models import init_chat_model

#llm = init_chat_model("anthropic:claude-sonnet-4-5-20250929", temperature=0.7)
llm = init_chat_model("google_genai:gemini-3.5-flash-lite", temperature=0.7)
response = llm.invoke(history)
response.pretty_print()
```
